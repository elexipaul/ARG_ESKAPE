## Import the files and merge them

```python
import pandas as pd
import glob
import warnings
warnings.filterwarnings('ignore')
file_pattern = '/home/user/Pranjal-11-12-2025/ESKAPE_ARG_Family/CARD_Results/ENT_rgi_results/*/*.txt'
file_paths = glob.glob(file_pattern)
if not file_paths:
    print("No files found. Please check the path spelling or file permissions.")
else:
    print(f"Successfully found {len(file_paths)} text files. Merging data...")       
    all_dataframes = [pd.read_csv(file, sep='\t') for file in file_paths]
    
    df1 = pd.concat(all_dataframes, ignore_index=True)       
    display(df1.head().style)
```

## Print the Unique drug classes 

```python
# Step 1: Drop NA values to prevent errors, then split the strings by the semicolon
# Example: "A; B" becomes ["A", " B"]
split_drugs = df1['Drug Class'].dropna().str.split(';')

# Step 2: Use .explode() to separate the lists into their own individual rows
exploded_drugs = split_drugs.explode()

# Step 3: Remove any accidental spaces at the beginning or end of the words
cleaned_drugs = exploded_drugs.str.strip()

# Step 4: Now that the redundancy is gone, recount the unique values!
true_drug_counts = cleaned_drugs.value_counts()

# Step 5: Display the clean, finalized data
print("True frequencies of individual drug classes:")
print(true_drug_counts)
```

## Frequency of AMR gene families

```python
split_drugs = df1['AMR Gene Family'].dropna().str.split(';')

# Step 2: Use .explode() to separate the lists into their own individual rows
exploded_drugs = split_drugs.explode()

# Step 3: Remove any accidental spaces at the beginning or end of the words
cleaned_drugs = exploded_drugs.str.strip()

# Step 4: Now that the redundancy is gone, recount the unique values!
true_drug_counts = cleaned_drugs.value_counts()

# Step 5: Display the clean, finalized data
print("True frequencies of individual AMR gene family:")
print(true_drug_counts)
```

```python
import os
import re

# ==========================================
# 1. DEFINE PARAMETERS
# ==========================================
output_dir = "/home/user/Pranjal-11-12-2025/ESKAPE_ARG_Family/Results/ENT_AMR_Gene_family/ARG_Fasta_Files"

# Define which column represents your "gene class" for the file names.
# You can change this to 'Drug Class' if you prefer.
class_column = 'AMR Gene Family'

# ==========================================
# 2. CREATE DIRECTORY
# ==========================================
os.makedirs(output_dir, exist_ok=True)

# ==========================================
# 3. CLEAN FILE NAMES & GROUP DATA
# ==========================================
# Create a temporary column that removes illegal file path characters (like /, \, ;, or spaces)
df1['Safe_Filename'] = df[class_column].astype(str).replace(r'[\\/*?:"<>|;]', "_", regex=True)

# Group the dataframe by this clean class name
grouped_df = df1.groupby('Safe_Filename')

# ==========================================
# 4. GENERATE FASTA FILES
# ==========================================
for gene_class, group in grouped_df:
    # Construct the full path for the FASTA file
    fasta_path = os.path.join(output_dir, f"{gene_class}.fasta")
    
    # Open the file in write mode
    with open(fasta_path, 'w') as fasta_file:
        
        # Iterate through every row in this specific gene class
        for index, row in group.iterrows():
            header = row['Best_Hit_ARO']
            sequence = row['Predicted_DNA']
            
            # Skip rows where the sequence might be missing (NaN)
            if pd.isna(sequence):
                continue
                
            # Write the header and sequence in standard FASTA format
            fasta_file.write(f">{header}\n")
            fasta_file.write(f"{sequence}\n")

print(f"Task complete! Check the '{output_dir}' directory for your FASTA files.")
```

```python
import os
import pandas as pd
import re

# ==========================================
# 1. DEFINE OUTPUT DIRECTORY
# ==========================================
metadata_dir = "/home/user/Pranjal-11-12-2025/ESKAPE_ARG_Family/Results/PA_AMR_Gene_family/Gene_Family_Metadata"
os.makedirs(metadata_dir, exist_ok=True)

# ==========================================
# 2. FILTER AND RENAME COLUMNS
# ==========================================
# Isolate only the columns we need from the main dataframe 'df'
columns_to_keep = ['Best_Hit_ARO', 'Start', 'Stop', 'Contig', 'Drug Class', 'AMR Gene Family']
metadata_df = df1[columns_to_keep].copy()

# Rename the columns to match your exact requested structure
metadata_df = metadata_df.rename(columns={
    'Best_Hit_ARO': 'AMR gene name',
    'Contig': 'Contig id',
    'Drug Class': 'Drug class'
})

# ==========================================
# 3. CLEAN FILE NAMES & GROUP DATA
# ==========================================
# Create a safe filename column by replacing illegal characters with underscores
metadata_df['Safe_Family_Name'] = metadata_df['AMR Gene Family'].astype(str).replace(r'[\\/*?:"<>|;]', "_", regex=True)

# Group the dataframe by the safe gene family name
grouped_metadata = metadata_df.groupby('Safe_Family_Name')

# ==========================================
# 4. EXPORT TO CSV
# ==========================================
# Define the exact column order you want in the final CSVs
final_csv_columns = ['AMR gene name', 'Start', 'Stop', 'Contig id', 'Drug class']

for family_name, group in grouped_metadata:
    # Filter the group to only include the final requested columns
    output_data = group[final_csv_columns]
    
    # Construct the file path and save
    csv_path = os.path.join(metadata_dir, f"{family_name}_metadata.csv")
    output_data.to_csv(csv_path, index=False)

print(f"Success! All metadata CSV files have been created in the '{metadata_dir}' directory.")
```
