## Import the files and merge them

```python
import pandas as pd
import glob
import warnings
warnings.filterwarnings('ignore')
file_pattern = '/home/user/Pranjal-11-12-2025/ESKAPE_ARG_Family/CARD_Results/Pa_rgi_results/*/*.txt'
file_paths = glob.glob(file_pattern)
if not file_paths:
    print("No files found. Please check the path spelling or file permissions.")
else:
    print(f"Successfully found {len(file_paths)} text files. Merging data...")       
    all_dataframes = [pd.read_csv(file, sep='\t') for file in file_paths]
    
    df1 = pd.concat(all_dataframes, ignore_index=True)       
    display(df1.head().style)
```

## Extraction of Actual ARG & Metadata

```python
import os
import pandas as pd

# ==========================================\n
# 1. DEFINE PATHS
# ==========================================\n
fasta_dir = "/home/user/Pranjal-11-12-2025/ESKAPE_ARG_Family/Results/Genefamily/Pa_ARG_Fasta_Files"
metadata_dir = "/home/user/Pranjal-11-12-2025/ESKAPE_ARG_Family/Results/Metadata/Pa_Gene_Family_Metadata"

os.makedirs(fasta_dir, exist_ok=True)
os.makedirs(metadata_dir, exist_ok=True)

# ==========================================\n
# 2. PREPARE AND CLEAN THE DATA
# ==========================================\n
# Isolate the exact columns we need from df1
columns_to_keep = ['Best_Hit_ARO', 'Predicted_DNA', 'Start', 'Stop', 'Contig', 'Drug Class', 'AMR Gene Family']
working_df = df1[columns_to_keep].copy()

# Critical Fix: Split the semicolons and EXPLODE into distinct rows
working_df['AMR Gene Family'] = working_df['AMR Gene Family'].dropna().str.split(';')
exploded_df = working_df.explode('AMR Gene Family')

# Strip any accidental whitespace from the separated names
exploded_df['AMR Gene Family'] = exploded_df['AMR Gene Family'].str.strip()

# ==========================================\n
# 3. CREATE SAFE FILENAMES AND GROUP
# ==========================================\n
# Replace any illegal file characters with an underscore
exploded_df['Safe_Family_Name'] = exploded_df['AMR Gene Family'].astype(str).replace(r'[\\/*?:"<>|]', "_", regex=True)

# Group the fully separated data. This will create exactly 50 groups!
grouped_data = exploded_df.groupby('Safe_Family_Name')

# ==========================================\n
# 4. GENERATE MATCHED FILES
# ==========================================\n
for family_name, group in grouped_data:
    
    # --- Generate the FASTA File ---
    fasta_path = os.path.join(fasta_dir, f"{family_name}.fasta")
    with open(fasta_path, 'w') as fasta_file:
        for _, row in group.iterrows():
            header = row['Best_Hit_ARO']
            sequence = row['Predicted_DNA']
            
            # Only write if a sequence exists
            if not pd.isna(sequence):
                fasta_file.write(f">{header}\n{sequence}\n")
                
    # --- Generate the Metadata CSV File ---
    csv_path = os.path.join(metadata_dir, f"{family_name}_metadata.csv")
    
    # Select just the metadata columns and rename them to your specifications
    meta_out = group[['Best_Hit_ARO', 'Start', 'Stop', 'Contig', 'Drug Class']].copy()
    meta_out = meta_out.rename(columns={
        'Best_Hit_ARO': 'AMR gene name',
        'Contig': 'Contig id',
        'Drug Class': 'Drug class'
    })
    
    # Save the CSV
    meta_out.to_csv(csv_path, index=False)

# ==========================================\n
# 5. VERIFICATION
# ==========================================\n
print(f"Task complete! Successfully processed {len(grouped_data)} unique AMR Gene Families.")
print(f"Generated {len(grouped_data)} FASTA files in: {fasta_dir}")
print(f"Generated {len(grouped_data)} Metadata files in: {metadata_dir}")
```

```python
# Define the folder path and map your files to the full organism names
folder_path = '/home/user/Pranjal-11-12-2025/ESKAPE_ARG_Family/Results/DRUG_class_family_mapping'
eskape_mapping = {
    'AB_drug_class_mapping.csv': 'Acinetobacter baumannii (AB)',
    'EF_drug_class_mapping.csv': 'Enterococcus faecium (EF)',
    'ENT_drug_class_mapping.csv': 'Enterobacter spp. (ENT)',
    'Kp_drug_class_mapping.csv': 'Klebsiella pneumoniae (Kp)',
    'Pa_drug_class_mapping.csv': 'Pseudomonas aeruginosa (Pa)',
    'SA_drug_class_mapping.csv': 'Staphylococcus aureus (SA)'
}

print("--- Step 1: Validating File Discovery ---")
# This will store only the files that are confirmed to exist
verified_files = {}

for file_name, org_name in eskape_mapping.items():
    file_path = os.path.join(folder_path, file_name)
    
    # Check if the file actually exists in the directory
    if os.path.exists(file_path):
        print(f"[SUCCESS] Discovered: {file_name} -> {org_name}")
        verified_files[file_name] = org_name
    else:
        print(f"[ERROR] Missing: {file_name}. Could not find it in '{folder_path}'.")
```

```python
# --- CELL 2: THE PROCESSING LOOP ---
print("--- Building the ESKAPE dictionary ---")

# CRITICAL: This is outside the loop, so it collects ALL data
presence_data = {}

for file_name, org_name in tqdm(verified_files.items(), colour='green', desc="Parsing Genomes"):
    file_path = os.path.join(folder_path, file_name)
    
    # Read the CSV
    df = pd.read_csv(file_path)
    
    # Clean, split by semicolon, and isolate unique families
    families = df['AMR Gene Family'].dropna().str.split(';').explode().str.strip()
    unique_families = set(families)
    
    # Map a '1' for every unique family found in this specific organism
    presence_data[org_name] = {family: 1 for family in unique_families}

# Print a check to ensure all 6 organisms are in memory
print(f"\n[INFO] Data successfully collected for: {list(presence_data.keys())}")
```

```python
# --- CELL 3: MATRIX GENERATION & EXPORT ---

# Convert the populated dictionary into a DataFrame matrix
matrix_df = pd.DataFrame.from_dict(presence_data, orient='index')

# Fill missing values (NaN) with 0 and convert the entire matrix to integers
matrix_df = matrix_df.fillna(0).astype(int)

# Sort the columns alphabetically for a cleaner, organized look
matrix_df = matrix_df.reindex(sorted(matrix_df.columns), axis=1)

# Save the final matrix to a new CSV in your current working directory
output_filename = '/home/user/Pranjal-11-12-2025/ESKAPE_ARG_Family/Results/DRUG_class_family_mapping/ESKAPE_AMR_Presence_Absence_Matrix.csv'
matrix_df.to_csv(output_filename)

print("--- Final Matrix Ready ---")
print(f"Organisms (Rows): {matrix_df.shape[0]}")
print(f"Gene Families (Columns): {matrix_df.shape[1]}")
print(f"Saved successfully as: {output_filename}")

# Optional: Display the first 5 columns of the matrix right here in the notebook
display(matrix_df.iloc[:, :5].head())
```
