```python
# Filter for rows matching BOTH the specific FASTA file AND an omega value > 1.0
filtered_df = df[(df['Input_FASTA'] == 'f108_merged.fasta') & (df['dN_dS_ML'] > 1.0)]

# To check how many rows are left after filtering
print(f"Filtered rows: {len(filtered_df)}")
filtered_df.head().style
```

```python
# Configuration: set the output CSV path before running this cell.
OUTPUT_CSV = '/home/user/Pranjal-11-12-2025/ESKAPE_ARG_Family/Results/selection_analysis/Family_dNdS_Summary.csv'

from pathlib import Path
from IPython.display import display

# Use the complete df loaded above; count query records in each family.
summary_rows = []
for input_fasta, family_df in df.groupby('Input_FASTA', sort=True):
    positive_df = family_df[family_df['dN_dS_ML'] > 1.0]
    summary_rows.append({
        'Input_FASTA': input_fasta,
        'Total No of genes in that family': len(family_df),
        'No of genes with dn/ds >1': len(positive_df),
        'Name of the Genes with dn/ds >1': '; '.join(positive_df['Query_ID'].astype(str)),
    })

family_summary_df = pd.DataFrame(summary_rows, columns=[
    'Input_FASTA',
    'Total No of genes in that family',
    'No of genes with dn/ds >1',
    'Name of the Genes with dn/ds >1',
])

output_path = Path(OUTPUT_CSV)
output_path.parent.mkdir(parents=True, exist_ok=True)
family_summary_df.to_csv(output_path, index=False)
print(f'Summary saved to: {output_path}')
display(family_summary_df)
```
