```python
import pandas as pd

# Load the SHAP features dataset
shap_df = pd.read_csv("/home/user/Pranjal-11-12-2025/ESKAPE_ARG_Family/CARD_DATA/card-data/refined_details/Final_Results/ML_on_features_ARG/results/family_classifiers/top_features_shap.csv")

# Define the exact target family you want to investigate
target_family = "APH(3')" 

# Filter the dataframe for the specific family
family_features = shap_df[shap_df['family'] == target_family].copy()

if family_features.empty:
    print(f"Family '{target_family}' not found in the dataset.")
else:
    # 1. Calculate the maximum SHAP value for this specific family
    max_shap_value = family_features['shap_importance'].max()
    
    # 2. Apply the mathematical safety threshold (Max SHAP <= 2.0)
    if max_shap_value > 2.0:
        print(f"WARNING: The model for '{target_family}' is statistically unstable.")
        print(f"Max SHAP value ({max_shap_value:.2f}) exceeds the 2.0 threshold.")
        print("These features likely represent overfitted statistical noise.")
    else:
        # 3. If the model is healthy, sort by importance and display the biological drivers
        family_features = family_features.sort_values(by='shap_importance', ascending=False)
        
        print(f"Model for '{target_family}' is stable (Max SHAP: {max_shap_value:.2f}).")
        print("Top structural drivers:")
        print(family_features[['feature', 'shap_importance']].head(20))
```

```python
import pandas as pd

# Load the card features dataset
card_df = pd.read_csv("/home/user/Pranjal-11-12-2025/ESKAPE_ARG_Family/CARD_DATA/card-data/card_reference_headers.csv")
card_df.head().style
```

```python
# Replace 'AMR Gene Family' with the exact name of your target column
target_column = 'organism'

# 1. Get the total NUMBER of unique values
unique_count = card_df[target_column].nunique()
print(f"Total unique values in {target_column}: {unique_count}")

# 2. Get the actual NAMES of the unique values
unique_names = card_df[target_column].unique()
print(f"\nList of unique names:\n{unique_names}")

# 3. Get the names AND the exact frequency of each occurrence 
value_frequencies = card_df[target_column].value_counts()
print(f"\nBreakdown of counts per unique value:\n{value_frequencies}")
```

```python
import os
output_txt_path = os.path.join('/home/user/Pranjal-11-12-2025/ESKAPE_ARG_Family/CARD_DATA/card-data/unique_amr_oraganisms_.txt')

# Open the text file in 'write' ('w') mode
with open(output_txt_path, 'w') as file:
    # Loop through each unique name in the array
    for name in unique_names:
        # Write the name to the file, followed by '\n' to push the next item to a new line
        file.write(f"{name}\n")

print(f"Successfully saved all unique names to: {output_txt_path}")
```
