```python
import os
import pandas as pd
from pathlib import Path
dir = Path("/home/user/Pranjal-11-12-2025/ESKAPE_ARG_Family/plasmid_study")
os.makedirs('/home/user/Pranjal-11-12-2025/ESKAPE_ARG_Family/plasmid_study/Refined_details', exist_ok=True)
```

```python
output_dir = dir / "Refined_details"
print(f"Directory ready at: {output_dir}")
```

```python
df1 = pd.read_csv(dir /'nuccore.csv')
df2 = pd.read_csv(dir /'taxonomy.csv')
df3 = pd.read_csv(dir /'typing.csv')
df4 = pd.read_csv(dir /'plasmidfinder.csv')
df5 = pd.read_csv(dir /'assembly.csv')
df6 = pd.read_csv(dir /'biosample.csv')
```

```python
# =============================================================================
# Step 1: Select only the necessary columns from each loaded DataFrame.
# Make sure to ALWAYS include the column you plan to use for joining!
# =============================================================================

# Example: df1 and df2 share the exact same key column named 'Accession'
df1_subset = df1[['NUCCORE_ACC', 'NUCCORE_UID', 'NUCCORE_Description' ,'NUCCORE_Length', 'TAXONOMY_UID', 'ASSEMBLY_UID' , 'BIOSAMPLE_UID' ]]
df2_subset = df2[['TAXONOMY_UID', 'TAXONOMY_taxon_name']]
df3_subset = df3[['relaxase_type(s)', 'NUCCORE_ACC', 'predicted_mobility' ,'PMLST_sequence_type', 'mash_neighbor_identification' ]]
df4_subset = df4[['NUCCORE_ACC', 'typing', 'organism_L2' ,'hit_id']]
df5_subset = df5[['ASSEMBLY_UID', 'ASSEMBLY_ACC']]
df6_subset = df6[['BIOSAMPLE_UID', 'ECOSYSTEM_query' , 'DISEASE_ecotag' , 'DISEASE_query' , 'DISEASE_ontid' , 'DISEASE_term' , 'DISEASE_group' ,'LOCATION_query', 'BIOSAMPLE_ACC' , 'ECOSYSTEM_tags']]
# =============================================================================
# Step 2: Merge the subsets step-by-step.
# We use how='left' to keep all rows from df1 and attach matching data.
# =============================================================================

# 1. Merge df1 and df2 (They share the exact same header: 'Accession')
final_df = df1_subset.merge(df2_subset, on='TAXONOMY_UID', how='left')

# 2. Merge with df3 (Different headers: 'Accession' in final_df, 'sample_id' in df3)
final_df = final_df.merge(df3_subset, on='NUCCORE_ACC',how='left')

# 3. Merge the remaining DataFrames (Assuming they share 'Accession')
final_df = final_df.merge(df4_subset, on='NUCCORE_ACC', how='left')
final_df = final_df.merge(df5_subset, on='ASSEMBLY_UID', how='left')
final_df = final_df.merge(df6_subset, on='BIOSAMPLE_UID', how='left')

# =============================================================================
# Step 3: Clean up and verify
# =============================================================================

# If you merged using different headers (like in step 2), you might have both 
# 'Accession' and 'sample_id' in your final table. You can drop the duplicate:
#final_df = final_df.drop(columns=['sample_id'])

# Check the final shape and the first few rows
print(f"Final merged dataset shape: {final_df.shape}")
final_df.tail(15)
```

```python
output_file_path = output_dir / "final_merged_metadata.csv"

# 3. Save the DataFrame to the CSV file
# index=False prevents pandas from writing row numbers into the file
final_df.to_csv(output_file_path, index=False)

print(f"Success! Dataset successfully saved to: {output_file_path}")
```
