```python
import pandas as pd
import glob
import warnings

warnings.filterwarnings('ignore')

# Step 1: Use the absolute path to point directly to your ESKAPE_ARG_Family directory
# The /*/*.txt grabs all text files inside the GCA_* subfolders
file_pattern = '/home/user/Pranjal-11-12-2025/ESKAPE_ARG_Family/Kp_rgi_results/*/*.txt'
file_paths = glob.glob(file_pattern)

# Step 2: Validate that files were found
if not file_paths:
    print("No files found. Please check the path spelling or file permissions.")
else:
    print(f"Successfully found {len(file_paths)} text files. Merging data...")
    
    # Step 3: Read and combine all files
    # Using a list comprehension here is cleaner and faster for genomic pipelines
    all_dataframes = [pd.read_csv(file, sep='\t') for file in file_paths]
    
    df = pd.concat(all_dataframes, ignore_index=True)
    
    # Step 4: Display the result
    display(df.head().style)
```

```python
unique_values = df['Drug Class'].unique()

# Step 2: Print the results nicely
print("The unique values in this column are:")
print(unique_values)
```

```python
# Step 1: Find unique values and count how often they appear
# Replace 'Best_Hit_ARO' with your actual column name
unique_counts = df['AMR Gene Family'].value_counts()

# Step 2: Print the results
print("Unique values and their frequencies:")
print(unique_counts)
```

```python
# Step 1: Drop NA values to prevent errors, then split the strings by the semicolon
# Example: "A; B" becomes ["A", " B"]
split_drugs = df['Drug Class'].dropna().str.split(';')

# Step 2: Use .explode() to separate the lists into their own individual rows
exploded_drugs = split_drugs.explode()

# Step 3: Remove any accidental spaces at the beginning or end of the words
cleaned_drugs = exploded_drugs.str.strip()

# Step 4: Now that the redundancy is gone, recount the unique values!
true_drug_counts = cleaned_drugs.value_counts()

# Step 5: Display the clean, finalized data
print("True frequencies of individual drug classes:")
print(true_drug_counts)
```

```python
# Step 1: Grab just the two columns we need and drop missing rows
# (Change 'AMR Gene Family' to 'Best_Hit_ARO' if you prefer specific gene names)
mapping_df = df[['Drug Class', 'AMR Gene Family']].dropna(subset=['Drug Class'])

# Step 2: Split the semicolon-separated text into lists
mapping_df['Drug Class'] = mapping_df['Drug Class'].str.split(';')

# Step 3: Explode the dataframe! This gives each drug its own row while keeping the gene attached
exploded_mapping = mapping_df.explode('Drug Class')

# Step 4: Strip away any extra spaces from the drug names
exploded_mapping['Drug Class'] = exploded_mapping['Drug Class'].str.strip()

# Step 5: Group the data by 'Drug Class', and collect the unique gene families for each
drug_class_to_genes = exploded_mapping.groupby('Drug Class')['AMR Gene Family'].unique()

# Step 6: Convert the result back into a clean, readable DataFrame
final_mapping = drug_class_to_genes.reset_index()

# Step 7: Display the first few rows to see your new mapping
final_mapping.head().style
```

```python
import os
# (Assuming pandas is already imported as pd from our previous steps)

# Step 1: Define your desired folder name and the CSV file name
target_folder = "/home/user/Pranjal-11-12-2025/ESKAPE_ARG_Family/Kp_rgi_results/Cleaned_RGI_Results"
file_name = "drug_class_mapping.csv"

# Step 2: Safely combine the folder and file name into a complete path
# Example output: Cleaned_RGI_Results/drug_class_mapping.csv
full_file_path = os.path.join(target_folder, file_name)

# Step 3: Create the directory
# exist_ok=True ensures the code doesn't crash if you run this cell twice
os.makedirs(target_folder, exist_ok=True)

# Step 4: Save the dataframe to the CSV file inside the new directory
# Replace 'df' with the specific dataframe variable you want to save (e.g., exploded_df or final_mapping)
mapping_df.to_csv(full_file_path, index=False)

# Step 5: Print a confirmation message so you know it worked!
print(f"Success! The directory '{target_folder}' was checked/created.")
print(f"Data successfully saved to: {full_file_path}")
```
