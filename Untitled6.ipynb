```python
import pandas as pd

# ── 1. Configuration ─────────────────────────────────────────────────────────
# Put the exact path to your CSV file here
CSV_FILE_PATH = "genomeNo_filtering_1.csv"

# ── 2. Load and Clean the Data ───────────────────────────────────────────────
try:
    # Read the CSV file
    df = pd.read_csv(CSV_FILE_PATH)
    
    # Ensure the columns exist before proceeding
    if 'NCBI' not in df.columns or 'BVBRC' not in df.columns:
        print("Error: Could not find 'NCBI' or 'BVBRC' columns in the file.")
        print(f"Columns found: {list(df.columns)}")
    else:
        # Drop empty values, convert to string, and remove extra whitespace
        ncbi_ids = df['NCBI'].dropna().astype(str).str.strip()
        bvbrc_ids = df['BVBRC'].dropna().astype(str).str.strip()
        
        # ── 3. Find the Overlap ──────────────────────────────────────────────
        # Convert the lists to sets for fast and easy comparison
        ncbi_set = set(ncbi_ids)
        bvbrc_set = set(bvbrc_ids)
        
        # Get the overlapping IDs (intersection)
        overlapping_ids = ncbi_set.intersection(bvbrc_set)
        
        # ── 4. Display the Results ───────────────────────────────────────────
        print(f"Total unique NCBI IDs : {len(ncbi_set):,}")
        print(f"Total unique BVBRC IDs: {len(bvbrc_set):,}")
        print(f"Number of overlapping IDs: {len(overlapping_ids):,}\n")
        
        if overlapping_ids:
            print("Here is a sample of the overlapping IDs:")
            # Convert set back to a list to slice the first 10 items
            for gen_id in list(overlapping_ids)[:14]: 
                print(f" - {gen_id}")
            
            if len(overlapping_ids) > 14:
                print(f"...and {len(overlapping_ids) - 14:,} more.")
                
            # Optional: Save the overlapping IDs to a new file for your records
            overlap_df = pd.DataFrame({'Overlapping_IDs': list(overlapping_ids)})
            overlap_df.to_csv("overlapping_genomes.csv", index=False)
            print("\nSaved the full list of overlapping IDs to 'overlapping_genomes.csv'")
            
        else:
            print("No overlapping IDs found between the two columns.")

except FileNotFoundError:
    print(f"Error: The file '{CSV_FILE_PATH}' was not found. Please check the spelling and path.")
```

```python
import pandas as pd

# ── 1. Configuration ─────────────────────────────────────────────────────────
# Put the exact path to your CSV file here
CSV_FILE_PATH = "genomeNo_filtering_1.csv"

# ── 2. Load and Clean the Data ───────────────────────────────────────────────
try:
    # Read the CSV file
    df = pd.read_csv(CSV_FILE_PATH)
    
    # Ensure the columns exist before proceeding
    if 'NCBI' not in df.columns or 'BVBRC' not in df.columns:
        print("Error: Could not find 'NCBI' or 'BVBRC' columns in the file.")
    else:
        # Drop empty values, convert to string, remove whitespace, and convert directly to sets
        ncbi_set = set(df['NCBI'].dropna().astype(str).str.strip())
        bvbrc_set = set(df['BVBRC'].dropna().astype(str).str.strip())
        
        # ── 3. Combine Everything (The Union) ────────────────────────────────
        # This combines the common IDs and the unique IDs into one master list
        complete_list = ncbi_set.union(bvbrc_set)
        
        # ── 4. Save to a Single Text File ────────────────────────────────────
        print("─── SAVING COMPLETE LIST ───")
        
        # Save the master list to a single text file
        with open("complete_id_list_2.txt", "w") as f_complete:
            for gen_id in sorted(complete_list):
                f_complete.write(f"{gen_id}\n")
                
        print(f"✓ Saved all {len(complete_list):,} unique IDs to 'complete_id_list.txt'")

except FileNotFoundError:
    print(f"Error: The file '{CSV_FILE_PATH}' was not found. Please check the spelling and path.")
```

```python
import pandas as pd
df = pd.read_csv("Isolates_Country.csv")
df.head().style
```

```python
col = df['Country']

# Number of unique values
print(f"Unique values count: {col.nunique()}")
# See the actual unique values
print(col.unique())
# Count each unique value
col.value_counts()
```

```python
col.value_counts()
```

```python
# Clean dataframe with counts
country_counts = col.value_counts().reset_index()
country_counts.columns = ['Country', 'Count']
country_counts = country_counts.reset_index(drop=True)

print(f"\nTotal unique countries: {len(country_counts)}")
country_counts
```

```python
country_counts.to_csv("country_counts.csv", index=False)
print("Saved successfully!")
```

```python
# Fix: convert to datetime first
df['Collection_date'] = pd.to_datetime(df['Collection_date'], errors='coerce')

# Check how many failed to parse
print(f"Unparseable dates (NaT): {df['Collection_date'].isna().sum()}")
print(f"Sample values: {df['Collection_date'].dropna().head()}")

country_counts = (df.groupby('Country')
                    .agg(
                        Count=('Country', 'count'),
                        Earliest_Date=('Collection_date', 'min'),
                        Latest_Date=('Collection_date', 'max')
                    )
                    .reset_index()
                    .sort_values('Count', ascending=False)
                    .reset_index(drop=True))

# Optional: format dates cleanly
country_counts['Earliest_Date'] = country_counts['Earliest_Date'].dt.strftime('%Y-%m-%d')
country_counts['Latest_Date'] = country_counts['Latest_Date'].dt.strftime('%Y-%m-%d')

country_counts
```

```python
country_counts.to_csv("country_counts_1.csv", index=False)
print("Saved successfully!")
```
